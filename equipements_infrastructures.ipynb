```python
codes_bizarres = []
for i in list(bpe_.columns):
    if i[:2] == "NB" : 
        codes_bizarres.append(i)

dict_bizarre = {} 
for c in codes_bizarres :
    dict_bizarre[c] = c[3:]
```

```python
# Dictionnaire pour stocker les groupes
# Clé = "Code_Année" (ex: 'C702_2015')
# Valeur = Liste des colonnes originales (ex: ['NB_C702PU_2015', 'NB_C702PR_2015'])
groupes = {}

for c in codes_bizarres:
    col_name = dict_bizarre[c]  # ex: 'NB_C702PU_2015'
    
    # 1. Nettoyage : on enlève le préfixe NB_ s'il existe pour l'analyse
    # (Si dict_bizarre contient déjà le nom sans NB, adaptez cette ligne)
    nom_analyse = col_name.replace("NB_", "") 
    
    # 2. Découpage pour trouver la clé racine
    parts = nom_analyse.split('_')
    
    # Sécurité : vérifier qu'on a bien assez de morceaux
    if len(parts) >= 2:
        racine_code = parts[0][:4]  # ex: 'C702' (Les 4 premiers caractères du premier morceau)
        annee = parts[-1]           # ex: '2015' (Le dernier morceau)
        
        # On forme la clé unique qui définit le groupe
        cle_groupe = f"{racine_code}_{annee}"
        
        # 3. Remplissage du dictionnaire
        if cle_groupe not in groupes:
            groupes[cle_groupe] = []
        groupes[cle_groupe].append(col_name)

# --- Résultat ---
# 'doublons' devient simplement la liste des valeurs de ce dictionnaire
# On ne garde que les groupes qui ont plus d'un élément (ceux qu'il faut sommer)
doublons = [liste_cols for liste_cols in groupes.values() if len(liste_cols) > 1]

# Affichage pour vérifier
doublons
```

```python
for groupe_cols in doublons:
    # 1. Définir le nom de la colonne finale (ex: C702_2015)
    # On prend le premier élément du groupe pour extraire les infos
    premier_elem = groupe_cols[0].replace("NB_", "")
    print(premier_elem)
    parts = premier_elem.split('_')
    racine = parts[0][:4]
    annee = parts[-1]
    nom_final = f"{racine}_{annee}" # Ou f"NB_{racine}_{annee}" selon votre préférence
    
    # 2. Créer la somme dans bpe_
    # bpe_[groupe_cols] sélectionne toutes les colonnes du groupe
    # .sum(axis=1) fait la somme ligne par ligne
    bpe_[nom_final] = bpe_[groupe_cols].sum(axis=1)
    
    # 3. (Optionnel) Supprimer les anciennes colonnes
    bpe_.drop(columns=groupe_cols, inplace=True)
```

```python
from bs4 import BeautifulSoup

# Chemin vers votre fichier
file_path = "/home/castro/Documents/ENSAE 2A/Stat'App/donnees/équipements/BPE24_liste_hierarchisee_TYPEQU.html"

def construire_dico_bpe(html_file):
    dictionnaire_equipements = {}
    
    with open(html_file, 'r', encoding='utf-8') as f:
        soup = BeautifulSoup(f, 'html.parser')
        
        # On récupère toutes les lignes du tableau
        rows = soup.find_all('tr')
        
        for row in rows:
            # On récupère toutes les cellules de la ligne
            cols = row.find_all('td')
            
            # On s'assure qu'il y a bien au moins 2 colonnes (Code et Libellé)
            if len(cols) >= 2:
                # .get_text(strip=True) permet d'enlever les espaces inutiles et balises internes
                code = cols[0].get_text(strip=True)
                libelle = cols[1].get_text(strip=True)
                
                # On évite les lignes vides éventuelles
                if code:
                    dictionnaire_equipements[code] = libelle
                    
    return dictionnaire_equipements

# Exécution
code_libelle_equipement = construire_dico_bpe(file_path)

# Vérification : Afficher les 5 premiers éléments
for k, v in list(code_libelle_equipement.items())[:5]:
    print(f"'{k}': '{v}'")
```

```python
import pandas as pd
import numpy as np
from joblib import Parallel, delayed

# 1. PRÉPARATION DES METADONNÉES
# On s'assure que les années sont triées
sorted_années = sorted(list(années))

# On génère la liste théorique de TOUTES les colonnes dans l'ordre exact :
# [Eq1_An1, Eq1_An2, ..., Eq2_An1, Eq2_An2, ...]
cols_ordered = [f"{e}_{a}" for e in eq_niv_3 for a in sorted_années]

# 2. PRÉPARATION DU DATAFRAME (VECTORISATION)
# On met DEPCOM en index pour aligner les données
if 'DEPCOM' in bpe_29.columns:
    df_work = bpe_29.set_index('DEPCOM')
else:
    df_work = bpe_29.copy()

# C'est l'étape clé : .reindex() va :
# - Garder uniquement les colonnes utiles
# - Les mettre dans le bon ordre
# - Créer automatiquement des NaN pour les colonnes manquantes (remplace tes `if nom_col in commune`)
# - Tout faire en C (très rapide)
df_work = df_work.reindex(columns=cols_ordered)

# 3. TRANSFORMATION EN MATRICE 3D (RESHAPE)
# Au lieu de boucler, on transforme tout le tableau 2D en un cube 3D d'un coup.
# Dimensions : (Nombre de Communes, Nombre d'Équipements, Nombre d'Années)
n_communes = len(df_work)
n_equips = len(eq_niv_3)
n_annees = len(sorted_années)

# .reshape est instantané et ne coûte quasiment pas de CPU
data_3d = df_work.values.reshape(n_communes, n_equips, n_annees)
index_communes = df_work.index.tolist()

# 4. CRÉATION PARALLÉLISÉE DU DICTIONNAIRE (UTILISATION CPU MAX)
# On définit une fonction qui emballe une "tranche" de la matrice dans un DataFrame
def create_commune_df(commune_id, data_slice):
    # data_slice est déjà une matrice numpy (Equip x Année), pas besoin de boucles
    return commune_id, pd.DataFrame(data_slice, index=eq_niv_3, columns=sorted_années)

# Joblib va lancer des processus sur tous tes cœurs (n_jobs=-1)
print(f"Lancement du traitement sur {n_communes} communes...")

results = Parallel(n_jobs=-1, backend='loky')(
    delayed(create_commune_df)(commune_id, data_3d[i]) 
    for i, commune_id in enumerate(index_communes)
)

# 5. ASSEMBLAGE FINAL
bpe_commune = dict(results)

print("Terminé !")
```

```python
import os
import matplotlib
import numpy as np
import concurrent.futures
from matplotlib.figure import Figure
from matplotlib.backends.backend_agg import FigureCanvasAgg
import time

# ==============================================================================
# CONFIGURATION
# ==============================================================================
OUTPUT_DIR = "/run/media/castro/SSD/bpe"
NUM_WORKERS = 30  # Laissez 1 ou 2 coeurs pour le système, mettez 32 pour tout utiliser.

# ==============================================================================
# FONCTION WORKER (Exécutée par chaque cœur CPU)
# ==============================================================================
def generer_graphes_commune(args):
    """
    Génère tous les graphiques pour une commune donnée.
    Args est un tuple pour compatibilité avec executor.map :
    (code_commune, df_commune, liste_equipements_cibles, dico_libelles, output_root)
    """
    depcom, df, eq_cibles, libelles, root_dir = args
    
    # 1. Création rapide du dossier
    # os.makedirs est thread-safe au niveau système de fichier
    dossier_target = os.path.join(root_dir, depcom)
    os.makedirs(dossier_target, exist_ok=True)
    
    # 2. Préparation des axes (X = Années)
    # df a les années en colonnes. On s'assure qu'elles sont triées.
    # On suppose que les colonnes sont des strings ou ints. On trie pour l'affichage.
    cols_annees = sorted(df.columns)
    df = df[cols_annees] # Réordonne le DF
    x_axis = [str(c) for c in cols_annees] # Labels X convertis en str
    
    compteur_img = 0
    
    # 3. Boucle sur les équipements (Lignes du DF)
    # On ne prend que ceux qui sont dans eq_cibles ET présents dans le DF
    equipements_a_traiter = [e for e in df.index if e in eq_cibles]
    
    for eq in equipements_a_traiter:
        serie = df.loc[eq]
        
        # Optimisation : On ne trace pas si tout est vide ou nul (gain de temps I/O)
        if serie.sum() > 0:
            
            # --- RENDU PUR (Sans plt.plot) ---
            # On instancie une Figure "légère"
            fig = Figure(figsize=(8, 5), dpi=100) 
            canvas = FigureCanvasAgg(fig)
            ax = fig.add_subplot(111)
            
            # Tracé
            # serie.values est un numpy array, très rapide à plotter
            ax.plot(x_axis, serie.values, marker='o', linestyle='-', color='#1f77b4', linewidth=1.5)
            
            # Textes
            titre_libelle = libelles.get(eq, eq)
            ax.set_title(f"{depcom} - {titre_libelle}", fontsize=10)
            ax.set_ylabel("Nombre")
            ax.grid(True, linestyle=':', alpha=0.6)
            
            # Rotation des dates si besoin (optionnel, coûte un peu de CPU)
            # ax.tick_params(axis='x', rotation=45)

            # --- SAUVEGARDE ---
            nom_fichier = f"{depcom}_{eq}.png"
            chemin_complet = os.path.join(dossier_target, nom_fichier)
            
            # savefig via le canvas est plus rapide que plt.savefig
            # bbox_inches='tight' est joli mais coûteux en CPU. Enlevez-le pour max perf.
            fig.savefig(chemin_complet, format='png') 
            
            compteur_img += 1
            
            # Nettoyage explicite (bien que le GC fasse le job, c'est plus sûr en multiprocessing)
            del fig
            del canvas

    return compteur_img

# ==============================================================================
# MAIN
# ==============================================================================
if __name__ == '__main__':
    # Vérification que bpe_commune existe (pour le contexte)
    # bpe_commune = ... (votre dictionnaire existant)
    # eq_niv_3 = ... (votre liste d'équipements à tracer)
    # code_libelle_equipement = ... (votre dictionnaire de titres)

    print(f"Préparation des tâches pour {len(bpe_commune)} communes...")
    
    if not os.path.exists(OUTPUT_DIR):
        os.makedirs(OUTPUT_DIR)

    # Transformation du dictionnaire en liste d'arguments pour le mapping
    # On passe eq_niv_3 converti en set pour une recherche O(1) dans le worker si la liste est longue
    set_eq_cibles = set(eq_niv_3)
    
    taches = []
    for depcom, df_ville in bpe_commune.items():
        # On crée le "kit" pour chaque worker
        taches.append((depcom, df_ville, set_eq_cibles, code_libelle_equipement, OUTPUT_DIR))

    print(f"Lancement sur {NUM_WORKERS} threads...")
    t0 = time.time()

    # ProcessPoolExecutor est vital ici (ThreadPoolExecutor serait bloqué par le GIL Python sur Matplotlib)
    total_graphes = 0
    with concurrent.futures.ProcessPoolExecutor(max_workers=NUM_WORKERS) as executor:
        # chunksize=50 signifie qu'on envoie les communes par paquets de 50 aux ouvriers
        # cela réduit l'overhead de communication inter-processus
        resultats = list(executor.map(generer_graphes_commune, taches, chunksize=50))

    total_graphes = sum(resultats)
    t1 = time.time()
    
    print("-" * 40)
    print(f"TERMINE en {t1 - t0:.2f} secondes.")
    print(f"Total images générées : {total_graphes}")
    print(f"Vitesse moyenne : {total_graphes / (t1 - t0):.0f} graphes/sec")
```

```python
import os
import matplotlib
import numpy as np
import concurrent.futures
from matplotlib.figure import Figure
from matplotlib.backends.backend_agg import FigureCanvasAgg
import time

# ==============================================================================
# CONFIGURATION
# ==============================================================================
# Le même chemin racine que tout à l'heure
OUTPUT_DIR = "/run/media/castro/SSD/bpe"
NUM_WORKERS = 30  # Puissance max

# ==============================================================================
# FONCTION WORKER (Niveau 1 - Agrégé)
# ==============================================================================
def generer_graphes_niv1_commune(args):
    depcom, df_niv1, libelles, root_dir = args
    
    # Construction du chemin : SSD/bpe/01001
    dossier_target = os.path.join(root_dir, depcom)
    os.makedirs(dossier_target, exist_ok=True)
    
    cols_annees = sorted(df_niv1.columns)
    df_niv1 = df_niv1[cols_annees]
    x_axis = [str(c) for c in cols_annees]
    
    compteur_img = 0
    
    for cat_lettre in df_niv1.index:
        serie = df_niv1.loc[cat_lettre]
        
        if serie.max() > 0:
            
            # --- CONFIG 500x500 ---
            fig = Figure(figsize=(5, 5), dpi=100, layout='tight')
            canvas = FigureCanvasAgg(fig)
            ax = fig.add_subplot(111)
            
            # Tracé
            ax.plot(x_axis, serie.values, marker='o', linestyle='-', color='#c0392b', linewidth=1.5)
            
            # Titres
            titre_cat = libelles.get(cat_lettre, f"Catégorie {cat_lettre}")
            ax.set_title(f"[{depcom}] {titre_cat}", fontsize=10, fontweight='bold')
            ax.set_ylabel("Total Catégorie")
            ax.grid(True, linestyle=':', alpha=0.6)
            
            # Rotation
            ax.tick_params(axis='x', rotation=45, labelsize=8)

            # --- SAUVEGARDE (JPG) ---
            nom_fichier = f"{depcom}_{cat_lettre}.jpg"
            chemin_complet = os.path.join(dossier_target, nom_fichier)
            
            # CORRECTION ICI : suppression de quality=90 qui faisait planter
            fig.savefig(chemin_complet, format='jpg')
            
            compteur_img += 1
            
            del fig
            del canvas

    return compteur_img

# ==============================================================================
# MAIN
# ==============================================================================
if __name__ == '__main__':
    # On vérifie que bpe_commune_niv1 est bien en mémoire
    print(f"Préparation des tâches Niv1 pour {len(bpe_commune_niv1)} communes...")
    
    if not os.path.exists(OUTPUT_DIR):
        os.makedirs(OUTPUT_DIR)

    taches_niv1 = []
    for depcom, df_ville_niv1 in bpe_commune_niv1.items():
        # On envoie les arguments au worker
        taches_niv1.append((depcom, df_ville_niv1, code_libelle_equipement, OUTPUT_DIR))

    print(f"Lancement du traitement Niv1 sur {NUM_WORKERS} threads...")
    t0 = time.time()

    total_graphes = 0
    with concurrent.futures.ProcessPoolExecutor(max_workers=NUM_WORKERS) as executor:
        # chunksize plus grand car traitement très rapide
        resultats = list(executor.map(generer_graphes_niv1_commune, taches_niv1, chunksize=200))

    total_graphes = sum(resultats)
    t1 = time.time()
    
    print("-" * 40)
    print(f"TERMINE en {t1 - t0:.2f} secondes.")
    print(f"Images générées : {total_graphes}")
    print(f"Destination : {OUTPUT_DIR}/[CodeCommune]/")
```

On va ici tracer les graphes en ligne brisée mais on divisera chaque colonne d'année de bpe_departement_niv1 par la population départementale correspondante à l'année donnée, de sorte à avoir le nombre d'équipements et leur évolution relative à l'évolution de population.

```python


# ---------------------------------------------------------------------------
# CRÉATION DE bpe_departement_niv1 
# (Somme géographique des catégories A, B, C... par Département)
# ---------------------------------------------------------------------------

print("1. Fusion de bpe_commune_niv1 en un Super-DataFrame...")
# On concatène le dictionnaire existant.
# Index résultant : (DEPCOM, CAT_NIV1)
df_all_niv1 = pd.concat(bpe_commune_niv1, names=['DEPCOM', 'CAT_NIV1'])

print("2. Identification des départements...")
# Fonction de gestion des DOM-TOM (971, 972...) vs Métropole (01, 75...)
def extract_dept(code_commune):
    s_code = str(code_commune)
    if s_code.startswith('97'):
        return s_code[:3]
    return s_code[:2]

# On applique la fonction sur l'index DEPCOM
index_communes = df_all_niv1.index.get_level_values('DEPCOM')
codes_departements = index_communes.map(extract_dept)

print("3. Sommation géographique par département...")
# On groupe par [Département, Catégorie] et on somme
# min_count=0 permet de conserver les NaN si aucune commune n'a de donnée pour une année
df_dept_niv1_groupe = df_all_niv1.groupby([codes_departements, 'CAT_NIV1']).sum(min_count=0)

print("4. Conversion en dictionnaire bpe_departement_niv1...")
bpe_departement_niv1 = {
    dep: data.droplevel(0) 
    for dep, data in df_dept_niv1_groupe.groupby(level=0)
}

# --- VÉRIFICATION ---
# Prenons un département au hasard pour vérifier
dep_test = list(bpe_departement_niv1.keys())[0]
print(f"\nStructure Niv1 pour le département {dep_test} :")
print(bpe_departement_niv1[dep_test].head(10))

# Nettoyage mémoire
del df_all_niv1, df_dept_niv1_groupe
```
